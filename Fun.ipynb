### 1: Ricerca giocatori in base al nome ###

### 6: Anno di inizio carriera di attività dei giocatori ###

```python
import csv
data = []   #definisci una lista vuota
def inizio_carriera(row):    #definisci una funzione chiamata "inizio_carriera" che prende come parametri le righe
    data.append({    #"appendo" alla lista le colonne che seguono
        'Player':row[0],    #"Player" nella colonna 0
        'From':int(row[1])})     #"From" nella colonna 1, cioè la data di inizio carriera
with open("player_data.csv", encoding="utf8") as f:     #apertura del file: "player_data.csv" e lettura del file codificato in "utf8", in assenza dell'ultimo blocco relativo alla codifica, il programma non andrà in esecuzione ed il file non sarà aperto
    reader = csv.reader(f,delimiter=",")    #leggo il file che ha come delimitatore la virgola
    for pos, riga in enumerate(reader):    #leggo il file
        if pos > 0:    #per ogni posizione maggiore di 0, cioè eludo l'intestazione
            inizio_carriera(riga)    #raccogli l'inizio della carriera
print(data)   #stampi i dati considerati
```

### 8: Giocatori giocanti per decade ###

```python
import codecs      #il modulo codecs consente di leggere il file codificato in "utf8"

def conta_periodi():    #viene definita la funzione "conta_periodi"

    f = open("player_data.csv", "r")      #apertura del file "player_data.csv" in modalità lettura
    f = codecs.open("player_data.csv", "r", "utf-8")    #attraverso il modulo "codecs.open" si apre il file "player_data.csv" in modalità lettura e codificato in utf8
    reader = csv.DictReader(f)      #Il metodo "DictReader" serve per leggere il file csv  
    data = []  #si inizializza una lista vuota
    for record in reader:    #per ogni elemento nel file
        data.append(record)    #appendo tutti i dati
    f.close()    #chiudo il file
print(len(data))    #stampo la lunghezza del dataset
print(data)   #stampo i dati sotto forma di dizionario, che mi serve per l'analisi successiva
```

### 16: Ricerca giocatori in base all'asterisco (hall of fame) ###
